```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 读数据 + 清洗列名
df = pd.read_csv('pakistan_social_media_usage_2026.csv')
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')

# 设置风格
sns.set_style("white")
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Scrolling Away Success: How Social Media Erodes Academic Performance', 
             fontsize=18, fontweight='bold', y=0.98)

# 面板1：社媒时长 vs 考试成绩
ax1 = axes[0, 0]
sns.regplot(data=df, x='daily_hours', y='exam_score', 
            ax=ax1, scatter_kws={'alpha':0.6, 'color':'#E76F51', 's':40},
            line_kws={'color':'#264653', 'linewidth':3})
ax1.set_xlabel('Daily Social Media (hours)', fontsize=12)
ax1.set_ylabel('Exam Score', fontsize=12)
ax1.set_title('More Screen Time → Lower Exam Scores', fontsize=14, fontweight='bold')

# 面板2：社媒时长 vs 学习时长
ax2 = axes[0, 1]
sns.regplot(data=df, x='daily_hours', y='study_hours', 
            ax=ax2, scatter_kws={'alpha':0.6, 'color':'#E76F51', 's':40},
            line_kws={'color':'#264653', 'linewidth':3})
ax2.set_xlabel('Daily Social Media (hours)', fontsize=12)
ax2.set_ylabel('Study Hours', fontsize=12)
ax2.set_title('Social Media Cannibalizes Study Time', fontsize=14, fontweight='bold')

# 面板3：社媒时长 vs 心理健康
ax3 = axes[1, 0]
sns.regplot(data=df, x='daily_hours', y='mental_health_rating', 
            ax=ax3, scatter_kws={'alpha':0.6, 'color':'#E76F51', 's':40},
            line_kws={'color':'#264653', 'linewidth':3})
ax3.set_xlabel('Daily Social Media (hours)', fontsize=12)
ax3.set_ylabel('Mental Health Rating (1-10)', fontsize=12)
ax3.set_title('Excessive Use Linked to Lower Well-being', fontsize=14, fontweight='bold')

# 面板4：学习时长 vs 考试成绩（正面对照）
ax4 = axes[1, 1]
sns.regplot(data=df, x='study_hours', y='exam_score', 
            ax=ax4, scatter_kws={'alpha':0.6, 'color':'#2A9D8F', 's':40},
            line_kws={'color':'#264653', 'linewidth':3})
ax4.set_xlabel('Study Hours', fontsize=12)
ax4.set_ylabel('Exam Score', fontsize=12)
ax4.set_title('More Studying → Higher Scores (Positive Control)', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.subplots_adjust(top=0.93)
plt.savefig('social_media_dashboard.png', dpi=300, bbox_inches='tight')
plt.show()
```
